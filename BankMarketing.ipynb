```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
x=df1.drop(['subscribed','ID'],axis=1)
```

```python
y=df1['subscribed']
```

```python
from sklearn.preprocessing import StandardScaler
std1=StandardScaler()
x_scaled=pd.DataFrame(std1.fit_transform(x),columns=x.columns)
```

```python
def vifcheck(x_scaled):
    vif=pd.DataFrame()
    vif['Features']=x_scaled.columns
    vif['Variance_inflation_factor']=[variance_inflation_factor(x_scaled.values,i) for i in range (x_scaled.shape[1])]
    print(vif)
```

```python
vifcheck(x_scaled)
```

```python
from imblearn.over_sampling import SMOTE
smt=SMOTE()
x_sampled,y_sampled=smt.fit_resample(x_scaled,y)
```

```python
y_sampled.value_counts()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
from sklearn.model_selection import train_test_split
maxAcc=0
maxRs=0
for i in range(1,400):
    x_train,x_test,y_train,y_test=train_test_split(x_sampled,y_sampled,test_size=.30,random_state=i)
    rf=RandomForestClassifier()
    rf.fit(x_train,y_train)
    pred=rf.predict(x_test)
    accuracy=accuracy_score(y_test,pred)
    if accuracy>maxAcc:
        maxAcc=accuracy
        maxRs=i
print(f'Best Accuracy for the sampled date is {maxAcc} and randomstate is{maxRs}')
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import train_test_split,cross_val_score
```

```python
def modelcheck(models,x,y):
    x1_train,x1_test,y1_train,y1_test=train_test_split(x,y,test_size=.30,random_state=rs)
        
    for model in models:
          
            print(f'Model:{model.__class__.__name__} with random_state={rs}')
            model.fit(x1_train,y1_train)
            pred_model=model.predict(x1_test)
            accuracy=accuracy_score(y1_test,pred_model)
            print(f'Accuracy:{accuracy}')
            print('Confusion_matrix:',confusion_matrix(y1_test,pred_model))
            print('Classification_report:',classification_report(y1_test,pred_model))
            model_cross_score=cross_val_score(model,x,y,cv=5)
            print(model_cross_score)
            print(model_cross_score.mean())
        
            difference= accuracy-model_cross_score.mean()
            print(f'Difference Between Accuracy and Cross_val_score mean: {difference}\n')
        
rf1=RandomForestClassifier()
dtc1=DecisionTreeClassifier()
knn1=KNeighborsClassifier()
et1=ExtraTreesClassifier()
ada1=AdaBoostClassifier(algorithm='SAMME')
gb1=GradientBoostingClassifier()
x=x_sampled
y=y_sampled
rs=304
models=[rf1,dtc1,knn1,et1,ada1,gb1]
modelcheck(models,x,y)
```

```python
x1_train,x1_test,y1_train,y1_test=train_test_split(x_sampled,y_sampled,test_size=.30,random_state=304)
```

```python
et1.fit(x1_train,y1_train)
pred_model=et1.predict(x1_test)
accuracy=accuracy_score(y1_test,pred_model)
print(f'Accuracy:{accuracy}')
print('Confusion_matrix:',confusion_matrix(y1_test,pred_model))
print('Classification_report:',classification_report(y1_test,pred_model))
model_cross_score=cross_val_score(et1,x_sampled,y_sampled,cv=5)
print(model_cross_score)
print(model_cross_score.mean())
        
difference= accuracy-model_cross_score.mean()
print(f'Difference Between Accuracy and Cross_val_score mean: {difference}\n')
```

```python
from sklearn.model_selection import GridSearchCV, train_test_split
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_features': ['auto', 'sqrt', 'log2'],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [False, True]
}

grid_search = GridSearchCV(estimator=et1, param_grid=param_grid, 
                           cv=5, n_jobs=-1, verbose=2, scoring='accuracy')
grid_search.fit(x1_train, y1_train)

print(f"Best parameters found: {grid_search.best_params_}")

best_etc = grid_search.best_estimator_

y1_pred = best_etc.predict(x1_test)

print(classification_report(y1_test, y1_pred))
```

```python
eth1=ExtraTreesClassifier(bootstrap=False, max_depth= None, max_features= 'sqrt', min_samples_leaf= 1, min_samples_split= 2, n_estimators= 100)
eth1.fit(x1_train,y1_train)
pred_eth1=eth1.predict(x1_test)
accuracy=accuracy_score(y1_test,pred_eth1)
print(f'Accuracy:{accuracy}')
print('Confusion_matrix:',confusion_matrix(y1_test,pred_eth1))
print('Classification_report:',classification_report(y1_test,pred_eth1))
model_cross_score=cross_val_score(eth1,x_sampled,y_sampled,cv=5)
print(model_cross_score)
print(model_cross_score.mean())
        
difference= accuracy-model_cross_score.mean()
```
