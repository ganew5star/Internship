```python
def calculate_good_scorer(rating, votes, num_items):
    # Normalized rating (out of 100)
    normalized_rating = rating * 20  # Assuming the rating is out of 5, convert it to out of 100
    # Normalized votes (using max votes to normalize)
    max_votes = df_train['VOTES'].max()
    normalized_votes = (votes / max_votes) * 100  # Scale votes to be out of 100
    # Normalize number of items
    max_items = df_train['Number_of_Items'].max()
    normalized_items = (num_items / max_items) * 100  # Scale number of items to be out of 100
    # Calculate the Good Scorer with weights
    score = 0.4 * normalized_votes + 0.3 * normalized_rating + 0.3 * normalized_items
    return score
df_train['Good_Scorer2'] = df_train.apply(lambda row: calculate_good_scorer(row['RATING'], row['VOTES'], row['Number_of_Items']), axis=1)

print(df_train)
```

```python
def extract_cuisines(cuisines):
    # Split and clean each item
    items = [item.strip().lower() for item in cuisines.split(',')]
    return items

# Extract unique cuisine items from the entire column
all_items = df_train['CUISINES'].apply(extract_cuisines).explode().unique()

# Print all unique items
print("Unique cuisine items:", all_items)

# Create a mapping dictionary from unique items (manual or automatic)
# Example automatic mapping (for demonstration; you may need to customize it)
cuisine_mapping = {item: item.title() for item in all_items}

# Apply the mapping function
def map_cuisines(cuisines):
    items = [item.strip().lower() for item in cuisines.split(',')]
    mapped_items = [cuisine_mapping.get(item, item) for item in items]
    return ', '.join(sorted(set(mapped_items)))

# Apply the mapping function to create a new column
df_train['Mapped_Cuisines'] = df_train['CUISINES'].apply(map_cuisines)

# Display the updated DataFrame
print(df_train)
```

```python

# Normalize names
def normalize_text(text):
    return text.strip().lower()

# Apply normalization
df_train['Normalized_CITY'] = df_train['CITY'].apply(normalize_text)
df_train['Normalized_LOCATION'] = df_train['LOCALITY'].apply(normalize_text)

# Find unique normalized names
unique_cities = df_train['Normalized_CITY'].unique()
unique_locations = df_train['Normalized_LOCATION'].unique()

# Create mapping dictionaries
city_mapping = {name: idx + 1 for idx, name in enumerate(unique_cities)}
location_mapping = {name: f'{city_mapping[df_train[df_train["Normalized_LOCATION"] == name].iloc[0]["Normalized_CITY"]]}_{idx + 1}' 
                    for idx, name in enumerate(unique_locations)}

# Apply mappings
df_train['City_Code'] = df_train['Normalized_CITY'].map(city_mapping)
df_train['Location_Code'] = df_train.apply(lambda row: f'{city_mapping[row["Normalized_CITY"]]}{list(location_mapping.keys()).index(row["Normalized_LOCATION"]) + 1}', axis=1)

# Display the results
print("Unique cities:", unique_cities)
print("Unique locations:", unique_locations)
print("City mapping:", city_mapping)
print("Location mapping:", location_mapping)
print(df_train[['CITY', 'LOCALITY', 'Normalized_CITY', 'Normalized_LOCATION', 'City_Code', 'Location_Code']])
```
