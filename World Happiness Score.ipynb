# Observation:
    Based on the dataset the following things observed:
        1.Dataset has 158 rows and 12 columns
        2.Dataset has 2 categorical and 10 numerical value features
        3.Dataset has zero null value present in it.
    Let us discuss about the features involved in the dataset:

    1.Countries -> It denote countries name of which score more happiness value.
    
    2.Region -> It also denote name of the region which score belongs to it.
    
    3.Happiness Rank -> It gives the integer value of ranking based on the happiness scores.
    
    4.Standard error -> It represents how much confident of the happiness score achieved, smaller standard error gives the more       confident of the value.
    
    5.Economy(GDP per Capita) -> Its gives the value of how much health and prosperity numbered in the countries economy.
    
    6.Family -> How much scores it gives the happiness when they are in the family united.
    
    7.Health(Life Expectancy) -> How many years the person lives according to healthcare systems.
    
    8.Freedom -> Freedom is based on how much they are happily living with their freedom in their life.
    
    9.Trust(Government Corruption) -> Based on which the government corruption raised or least, if least then score give more         confident in trust of the government.
    
    10.Generosity -> The term generosity represent the citizens who donate times and money to the country happiness.
    
    11.Dystopia Residual -> This is the term of which gives the score of benchmark based on the country happiness.
    
    12.Happiness Score -> This is the outcomes and target feature of the dataset based on other features scored.
    
    From the DataSet the we going to analyse and predict the happiness score based on the above numerical features. Hence this       outcomes denotes Regression and not classification.

# Models

# Scenario:1 Removing Outliers and multicollinearity(Standard error and Happiness Rank)

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
maxRs=0
max_Acc=0
for i in range(1,200):
    x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.30,random_state=i)
    rf=RandomForestRegressor()
    rf.fit(x_train,y_train)
    pred=rf.predict(x_test)
    acc=r2_score(y_test,pred)
    error=np.sqrt(mean_squared_error(y_test,pred))
    if acc>max_Acc:
        maxRs=i
        max_Acc=acc
print(f'Best score{max_Acc} and least error{error} in randomstate{maxRs}')
```

```python
x1_train,x1_test,y1_train,y1_test=train_test_split(x,y,test_size=.30,random_state=161)
```

# Model: Linear Regression

```python
pred_lr1=lr1.predict(x1_test)
```

```python
print('r2_score:',r2_score(y1_test,pred_lr1))
print('mean_squared_error:',mean_squared_error(y1_test,pred_lr1))
print('RMSE:',np.sqrt(mean_squared_error(y1_test,pred_lr1)))
```

```python
from sklearn.model_selection import cross_val_score
```

```python
lr1_cross_score=cross_val_score(lr1,x,y,cv=5)
print(lr1_cross_score)
print(lr1_cross_score.mean())
print('Difference between accuracy and cross val:',((r2_score(y1_test,pred_lr1))-(lr1_cross_score.mean())))
```
