```python
import random
import pandas as pd
import numpy as np

# List of sample words to create sentences
words = ["this", "is", "a", "sample", "sentence", "with", "some", "random", "words"]

# Function to introduce spelling errors in a sentence
def introduce_spelling_errors(sentence, error_rate=0.1):
    words = sentence.split()
    num_errors = int(len(words) * error_rate)
    error_indices = random.sample(range(len(words)), num_errors)
    for i in error_indices:
        word = words[i]
        if len(word) > 1:
            swap_idx = random.randint(0, len(word) - 2)
            words[i] = word[:swap_idx] + word[swap_idx+1] + word[swap_idx] + word[swap_idx+2:]
    return ' '.join(words), len(error_indices)

# Function to introduce grammar errors in a sentence
def introduce_grammar_errors(sentence, error_rate=0.1):
    words = sentence.split()
    num_errors = int(len(words) * error_rate)
    error_indices = random.sample(range(len(words)), num_errors)
    for i in error_indices:
        if i < len(words) - 1:
            words[i], words[i+1] = words[i+1], words[i]
    return ' '.join(words)

# Generate random sentences
def generate_random_sentence():
    return ' '.join(random.choices(words, k=random.randint(5, 10)))

# Create the dataset
data = []
num_sentences = 1000

for _ in range(num_sentences):
    sentence = generate_random_sentence()
    
    spell_error = random.choice([0, 1])
    grammar_error = random.choice([0, 1])
    
    if spell_error:
        sentence, spell_count = introduce_spelling_errors(sentence)
    else:
        spell_count = 0
    
    if grammar_error:
        sentence = introduce_grammar_errors(sentence)
    
    data.append([sentence, spell_error, grammar_error, spell_count])

# Convert to DataFrame
df1 = pd.DataFrame(data, columns=['sentence', 'spelling_label', 'grammar_label', 'no_of_spell_errors'])

# Save to CSV
df1.to_csv('specialdataset.csv', index=False)

print("specialdataset.csv'.")
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

# Text preprocessing function
def preprocess_text(sentence):
    # Tokenization
    words = word_tokenize(sentence)
    # Lowercasing
    words = [word.lower() for word in words]
    # Removing punctuation and special characters
    words = [word for word in words if word.isalnum()]
    # Removing stopwords
    stop_words = set(stopwords.words('english'))
    words = [word for word in words if word not in stop_words]
    # Lemmatization
    lemmatizer = WordNetLemmatizer()
    words = [lemmatizer.lemmatize(word) for word in words]
    return ' '.join(words)

# Apply preprocessing
df['cleaned_sentence'] = df['sentences'].apply(preprocess_text)
print(df[['sentences', 'cleaned_sentence']])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# Split the data into training and testing sets
X_train, X_test, y_train_spelling, y_test_spelling, y_train_grammar, y_test_grammar = train_test_split(
    X, y_spelling, y_grammar, test_size=0.2, random_state=42)

# Train a Logistic Regression model for spelling error detection
model_spelling = LogisticRegression()
model_spelling.fit(X_train, y_train_spelling)

# Train a Logistic Regression model for grammar error detection
model_grammar = LogisticRegression()
model_grammar.fit(X_train, y_train_grammar)

# Predict on the test set
y_pred_spelling = model_spelling.predict(X_test)
y_pred_grammar = model_grammar.predict(X_test)

# Evaluate the models
print("Spelling Error Detection:")
print(classification_report(y_test_spelling, y_pred_spelling))
print("Accuracy:", accuracy_score(y_test_spelling, y_pred_spelling))

print("Grammar Error Detection:")
print(classification_report(y_test_grammar, y_pred_grammar))
print("Accuracy:", accuracy_score(y_test_grammar, y_pred_grammar))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report

# Correct the labels based on feedback
corrected_data = pd.DataFrame({
    "sentences": [
        "He plays the piano beautifully.",
        "We visited their grandparent's house last summer.",
        "She wrote a poem for her teacher.",
        "I have an apple in my bag.",
        "They're going to the park next weekend.",
        "He is going to the store.",
        "I saw that movie yesterday.",
        "She wrote a letter to her friend.",
        "They are visiting their home.",
        "She does not like how to write the novels."
    ],
    "grammar label": [
        0,  # Corrected
        0,  # Corrected
        0,  # Corrected
        0,  # No error
        0,  # No error
        0,  # Corrected
        0,  # Corrected
        0,  # Corrected
        0,  # Corrected
        0   # Corrected
    ],
    "spelling label": [
        0,  # Corrected
        0,  # Corrected
        0,  # Corrected
        0,  # No error
        0,  # No error
        0,  # No error
        0,  # No error
        0,  # No error
        0,  # No error
        0   # No error
    ]
})

# Split the dataset into training and test sets
sentences_train, sentences_test, labels_grammar_train, labels_grammar_test, labels_spelling_train, labels_spelling_test = train_test_split(
    corrected_data['sentences'], corrected_data['grammar label'], corrected_data['spelling label'], test_size=0.2, random_state=42)

# Vectorize the sentences using TF-IDF
vectorizer = TfidfVectorizer()
X_train = vectorizer.fit_transform(sentences_train)
X_test = vectorizer.transform(sentences_test)

# Define the parameter grid for RandomizedSearchCV
param_grid = {
    'n_estimators': [100, 200, 300, 400, 500],
    'max_features': ['auto', 'sqrt', 'log2'],
    'max_depth': [10, 20, 30, 40, 50, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True, False]
}

# Initialize the RandomForestClassifier
rf_classifier = RandomForestClassifier()

# Initialize RandomizedSearchCV for grammar errors
random_search_grammar = RandomizedSearchCV(
    estimator=rf_classifier, param_distributions=param_grid,
    n_iter=100, cv=3, verbose=2, random_state=42, n_jobs=-1
)

# Fit the model for grammar errors
random_search_grammar.fit(X_train, labels_grammar_train)

# Get the best parameters for grammar error detection
best_params_grammar = random_search_grammar.best_params_

# Initialize RandomizedSearchCV for spelling errors
random_search_spelling = RandomizedSearchCV(
    estimator=rf_classifier, param_distributions=param_grid,
    n_iter=100, cv=3, verbose=2, random_state=42, n_jobs=-1
)

# Fit the model for spelling errors
random_search_spelling.fit(X_train, labels_spelling_train)

# Get the best parameters for spelling error detection
best_params_spelling = random_search_spelling.best_params_

# Evaluate the grammar error model
predictions_grammar = random_search_grammar.predict(X_test)
grammar_report = classification_report(labels_grammar_test, predictions_grammar)

# Evaluate the spelling error model
predictions_spelling = random_search_spelling.predict(X_test)
spelling_report = classification_report(labels_spelling_test, predictions_spelling)

print("Best parameters for grammar error detection:", best_params_grammar)
print("Classification report for grammar error detection:\n", grammar_report)
print("Best parameters for spelling error detection:", best_params_spelling)
print("Classification report for spelling error detection:\n", spelling_report)
```
